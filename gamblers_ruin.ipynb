```python
import numpy as np
rng = np.random.default_rng()

money = 10      # start with £10
flips = 0
while money > 0 and money < 20:   # stop if broke (£0) or if we hit £20
    if rng.random() < 0.5:        # 50% chance to win this flip
        money += 1                # win £1
    else:
        money -= 1                # lose £1
    flips += 1

if money == 0:
    print("Went BROKE after", flips, "flips")
else:
    print("Reached £20 after", flips, "flips")
```

```python
import numpy as np
rng = np.random.default_rng()

broke_count = 0
n_gamblers = 10000

for gambler in range(n_gamblers):     # play the whole game 10,000 times
    money = 10
    while money > 0 and money < 100:   # one gambler's game
        if rng.random() < 0.5:
            money += 1
        else:
            money -= 1
    if money == 0:                    # did this gambler end broke?
        broke_count += 1

print(broke_count, "out of", n_gamblers, "went broke")
print("That's", 100 * broke_count / n_gamblers, "%")
```
